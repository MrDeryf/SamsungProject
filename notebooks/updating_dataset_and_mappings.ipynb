## Тут будет обновление набора данных

План:
1. Загрузить all_images.csv через пандас
2. Пройтись по каждой строке, заменить массив в поле styles на одно случайное значение из него (random.choice(row["styles"]))
3. Сохранить новый датафрейм с тем же названием (all_images.csv)
4. Сделать аналогичные действия с train.csv, val.csv, test.csv
5. Посмотреть, что всё норм сделалось

```python
import pandas as pd
import random
import ast
import shutil
import os
from pathlib import Path


# === НАСТРОЙКИ ===
FILE_TO_PROCESS = Path("../data/all_images.csv")
BACKUP_NAME = Path("../data/all_csv_copy.csv")
# =================

# 0. Создаём резервную копию перед изменениями
if os.path.exists(FILE_TO_PROCESS):
    shutil.copy(FILE_TO_PROCESS, BACKUP_NAME)
    print(f"Резервная копия сохранена: {BACKUP_NAME}")
else:
    print(f"Ошибка: файл {FILE_TO_PROCESS} не найден в папке со скриптом.")
    exit()

print(f"Загружаем {FILE_TO_PROCESS}...")
df = pd.read_csv(FILE_TO_PROCESS)

if 'styles' not in df.columns:
    print(" Колонка 'styles' не найдена. Проверьте заголовки CSV.")
else:
    print("Обрабатываем колонку 'styles'...")
    for idx in df.index:
        cell = df.at[idx, 'styles']

        # Пропускаем пустые ячейки
        if pd.isna(cell):
            continue

        # Преобразуем строковое представление списка в настоящий список
        if isinstance(cell, str):
            try:
                cell = ast.literal_eval(cell)
            except (ValueError, SyntaxError):
                # Фоллбэк: убираем скобки/кавычки, бьём по запятым
                cell = [item.strip().strip("'\"") for item in cell.replace('[', '').replace(']', '').split(',') if item.strip()]

        # Если это список и в нём есть элементы — оставляем один случайный
        if isinstance(cell, list) and len(cell) > 0:
            df.at[idx, 'styles'] = random.choice(cell)

# 3. Сохраняем обратно в тот же файл
df.to_csv(FILE_TO_PROCESS, index=False)
print(f"Изменения сохранены в {FILE_TO_PROCESS}")

# === 5. ПРОВЕРКА ===
print("\n ПРОВЕРКА РЕЗУЛЬТАТА:")
df_check = pd.read_csv(FILE_TO_PROCESS)
print(f"Всего строк: {len(df_check)}")
if 'styles' in df_check.columns:
    print(" Первые 5 значений в 'styles' после обработки:")
    for val in df_check['styles'].head(5):
        print(f"  - {val} (тип: {type(val).__name__})")
print(" Готово!")
```

## Тут будет обновление маппингов

План:
1. Вставить путь к папке mappings ниже
2. Добавить недостающие импорты
3. Создать копию папки mappings, сохранить её куда-нибудь
4. Выполнить две функции для all_images.csv
5. Сравнить новый маппинг с сохранённой копией


```python
import json
from pathlib import Path


PROCESSED_DATA_DIR = Path("../data/mappings")

```

```python
def create_artist_mappings(train_df):
    """
    Создаёт маппинг художник <-> ID на основе тренировочной выборки.
    """
    artists = sorted(
        [str(name).strip().lower() for name in train_df["artist"].unique()]
    )

    artist2id = {artist: idx for idx, artist in enumerate(artists)}
    id2artist = {idx: artist for artist, idx in artist2id.items()}

    PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)

    with open(PROCESSED_DATA_DIR / "artist2id.json", "w", encoding="utf-8") as f:
        json.dump(artist2id, f, ensure_ascii=False, indent=2)
    with open(PROCESSED_DATA_DIR / "id2artist.json", "w", encoding="utf-8") as f:
        json.dump(id2artist, f, ensure_ascii=False, indent=2)

    print(f"🎨 Создано {len(artists)} классов художников.")
    return artist2id, id2artist


def create_era_and_style_mappings(metadata_df):
    """
    Создаёт маппинги для стилей и эпох.
    """
    # Так как styles теперь не массив, а строка, просто берем уникальные значения
    # dropna() убирает пустые значения, astype(str) гарантирует тип, strip() убирает лишние пробелы
    all_styles = set(metadata_df["styles"].dropna().astype(str).str.strip())
    
    # Сортируем и создаем словарь style -> id
    style2id = {style: idx for idx, style in enumerate(sorted(all_styles))}

    # Для эпох код остался прежним
    all_eras = metadata_df["era"].dropna().unique().tolist()
    era2id = {era: idx for idx, era in enumerate(sorted(all_eras))}

    with open(PROCESSED_DATA_DIR / "style2id.json", "w", encoding="utf-8") as f:
        json.dump(style2id, f, ensure_ascii=False, indent=2)
    with open(PROCESSED_DATA_DIR / "era2id.json", "w", encoding="utf-8") as f:
        json.dump(era2id, f, ensure_ascii=False, indent=2)

    print(f"✅ Создано {len(style2id)} классов стилей и {len(era2id)} классов эпох.")
    return style2id, era2id
```

```python
import pandas as pd

all_images_df = pd.read_csv("../data/all_images.csv")


artist2id, id2artist = create_artist_mappings(all_images_df)

style2id, era2id = create_era_and_style_mappings(all_images_df)

print("\n Все маппинги успешно созданы и сохранены!")
```
